```python
# Import the data
import numpy as np
import pandas as pd
from modules.data_loading import load_participant_details
from config import RAW_PARTICIPANT_DETAILS_PATH, RAW_DATA_DIR

participant_details = load_participant_details(RAW_PARTICIPANT_DETAILS_PATH)
```

```python
# print the length of each participant's data in hours
lengths = []
for participant in participant_details:
    path = participant.filename
    if participant.rtsa_side is None:
        continue
    data = np.loadtxt(f'{RAW_DATA_DIR}/{path}', delimiter='\t', skiprows=1, usecols=range(1,2))
    length = data.shape[0] / 10 / 3600
    lengths.append(
        {
            "participant": participant.filename,
            "duration (hrs)": length
        }
    )

df = pd.DataFrame(lengths)
print(
    df.sort_values(by="duration (hrs)", ascending=False)
    .reset_index(drop=True)
    .round(2)
    .to_markdown(index=False)
)
```

Data length looks good!
